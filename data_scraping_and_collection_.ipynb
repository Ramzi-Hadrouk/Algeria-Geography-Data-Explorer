```python
import pandas as pd
import json
```

```python
table=pd.read_html('https://ar.wikipedia.org/wiki/%D9%88%D9%84%D8%A7%D9%8A%D8%A7%D8%AA_%D8%A7%D9%84%D8%AC%D8%B2%D8%A7%D8%A6%D8%B1',match='الخريطة')
```

```python
print(f'Total tables: {len(table)}')
```

```python
df = table[0]
df.head(30)
```

```python
print(df.info())
```

```python
last_row=df.tail(1)
df = df.iloc[:-1, :] 
```

```python
last_row.head()
```


<H1 style="color:#64ff33; font-size:34px;background:#222 ; border-radius:15px ; text-align:center ;padding:20px">Cities Only</H1>


## Extract Arabic city names ("Wilayat") into a single list


```python
# Extract Arabic city names ("Wilayat") into a single list
wilayat_ar = [row[3] for row in df.itertuples(index=False)]

print(wilayat_ar )
print(len(wilayat_ar) )

```


## Extract Latin city names ("Wilayat") into a single list

```python
# Extract Latin city names ("Wilayat") into a single list
wilayat_en = [row[2] for row in df.itertuples(index=False)]

print(wilayat_en )
print(len(wilayat_en))
```

## Export Cities as json files

```python
with open('wilayat_ar.json', 'w', encoding='utf-8') as f:
    json.dump(wilayat_ar, f, ensure_ascii=False)
```

```python
with open('wilayat_en.json', 'w', encoding='utf-8') as f:
    json.dump(wilayat_en, f, ensure_ascii=False)
```

<H1 style="color:#64ff33; font-size:34px;background:#222 ; border-radius:15px ; text-align:center ;padding:20px"> Cities  with Full inforation</H1>

```python
wilayat_full_information= {}
for row in df.itertuples(index=False):
    wilayat_full_information[int(row[1])]={"name_en":row[2] ,"name_ar":row[3] ,
                                      "number_of_dayrat":int(row[4]) ,"number_of_baladiyat":int(row[5]),
                                      "area_km":int(row[6]) ,"area_ml":int(row[7]) ,"population_2008":int(row[8] )}
```

```python
with open('wilayat_full_information.json', 'w', encoding='utf-8') as f:
    json.dump(wilayat_full_information, f, ensure_ascii=False)
```
