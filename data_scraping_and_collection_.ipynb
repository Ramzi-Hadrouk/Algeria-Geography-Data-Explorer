```python
import pandas as pd
import json
```

```python
table=pd.read_html('https://ar.wikipedia.org/wiki/%D9%88%D9%84%D8%A7%D9%8A%D8%A7%D8%AA_%D8%A7%D9%84%D8%AC%D8%B2%D8%A7%D8%A6%D8%B1',match='الخريطة')
```

```python
print(f'Total tables: {len(table)}')
```

```python
df = table[0]
df.head(30)
```

```python
print(df.info())
```

```python
last_row=df.tail(1)
df = df.iloc[:-1, :] 
```

# Cities


## Extract Arabic city names ("Wilayat") into a single list


```python
# Extract Arabic city names ("Wilayat") into a single list
wilayat_ar = [row[3] for row in df.itertuples(index=False)]

print(wilayat_ar )
print(len(wilayat_ar) )

```


## Extract Latin city names ("Wilayat") into a single list

```python
# Extract Latin city names ("Wilayat") into a single list
wilayat_en = [row[2] for row in df.itertuples(index=False)]

print(wilayat_en )
print(len(wilayat_en))
```

## Export Cities as json files

```python
with open('wilayat_ar.json', 'w', encoding='utf-8') as f:
    json.dump(wilayat_ar, f, ensure_ascii=False)
```

```python
with open('wilayat_en.json', 'w', encoding='utf-8') as f:
    json.dump(wilayat_en, f, ensure_ascii=False)
```
